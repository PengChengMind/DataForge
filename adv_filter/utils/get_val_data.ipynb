```python
import random
import os

def select_and_remove_random_lines(file1, file2, output_file, num_lines=400):
    # 读取文件内容
    with open(file1, 'r', encoding='utf-8') as f:
        lines1 = f.readlines()

    with open(file2, 'r', encoding='utf-8') as f:
        lines2 = f.readlines()

    # 从每个文件中随机挑选指定数量的行
    selected_lines1 = random.sample(lines1, min(num_lines, len(lines1)))
    selected_lines2 = random.sample(lines2, min(num_lines, len(lines2)))

    # 创建输出文件
    with open(output_file, 'w', encoding='utf-8') as f:
        f.writelines(selected_lines1)
        f.writelines(selected_lines2)

    # 从原文件中删除已选择的行
    remaining_lines1 = [line for line in lines1 if line not in selected_lines1]
    remaining_lines2 = [line for line in lines2 if line not in selected_lines2]

    with open(file1, 'w', encoding='utf-8') as f:
        f.writelines(remaining_lines1)

    with open(file2, 'w', encoding='utf-8') as f:
        f.writelines(remaining_lines2)

```

```python

# 文件路径
file1_path = '/home/zlj/code/pcl/template/data/dealdata/negative.txt'
file2_path = '/home/zlj/code/pcl/template/data/dealdata/positive.txt'
output_file_path = '/home/zlj/code/pcl/template/data/train-val_dataset/test_data/test500MIX.txt'

select_and_remove_random_lines(file1_path, file2_path, output_file_path, num_lines=400)

```
