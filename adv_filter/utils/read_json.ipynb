```python
import json


def read_data(input_file_path, output_file_path):
    print("------------------------")
    print("文件名：{}",input_file_path)
    data_filter = []
    seen = set()



    with open(input_file_path, 'r') as file, open(output_file_path, 'w') as output:
        for i, line in enumerate(file):
            if line.find("text") != -1: #过滤掉非text的文本
                if line not in seen:
                    seen.add(line)
        for i in seen:
            text_json = json.loads(i)
            test_list = text_json["text"].replace('\\n','').replace('\n','')+'\n'
            output.write(test_list)
    return data_filter
```

```python
import json


def read_hg_data(input_file_path, output_file_path):
    print("------------------------")
    print("文件名：{}",input_file_path)
    data_filter = []
    seen = set()
    total_lines = 0
    skip_lines = 0
    
    with open(input_file_path, 'r') as file, open(output_file_path, 'w') as output:
        for i, content in enumerate(file):
            text_json = json.loads(content)
            for line in text_json:
                label = '###0' if line['output'] == 'T' else '###1' 
                test_list = line['input']['chat'].replace('\\n','').replace('\n','')+ label+'\n'
                output.write(test_list)

    return data_filter
```
