```python
# 混入题目，随机
import random

def insert_news_titles_with_length_check(file_A_path, file_B_path, output_file_path, file_D_path):
    # 读取文件B中的新闻标题
    with open(file_B_path, 'r', encoding='utf-8') as file_B:
        news_titles = [line.strip() for line in file_B.readlines() if line.strip()]

    # 读取文件A中的所有行
    with open(file_A_path, 'r', encoding='utf-8') as file_A:
        lines_A = file_A.readlines()

    # 如果行数超过十万行，随机选择十万行
    if len(lines_A) > 100000:
        lines_A = random.sample(lines_A, 100000)

    with open(output_file_path, 'w', encoding='utf-8') as output_file, open(file_D_path, 'w', encoding='utf-8') as file_D:
        for line in lines_A:
            line = line.strip()  # 移除每行开头和结尾的空格
            
            if len(line) > 100:
                # 随机选择3-8条新闻标题
                random_titles = random.sample(news_titles, random.randint(3, 8))
                inserted_text = ' '.join(random_titles)

                # 随机选择插入的位置：头部、中间或尾部
                position = random.choice(['start', 'middle', 'end'])
                if position == 'start':
                    new_line = inserted_text + ' ' + line
                elif position == 'middle':
                    middle_index = len(line) // 2
                    new_line = line[:middle_index] + ' ' + inserted_text + ' ' + line[middle_index:]
                else:
                    new_line = line + ' ' + inserted_text

                # 将修改后的行写入输出文件
                output_file.write(new_line + "###0\n")
            else:
                # 将不超过100个字符的行写入文件D
                file_D.write(line + '\n')

# 使用示例
file_A = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/0data_source.txt'
file_B = '/home/zlj/code/pcl/template/data/keywords/新闻标题/headline.txt'
output_file = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1headline_output.txt'
file_D = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1headline_short.txt'

insert_news_titles_with_length_check(file_A, file_B, output_file, file_D)
```

```python
def search_keywords(text_file_path, keyword_file_path, output_file_path, get_match=True):
    # 读取关键词库
    with open(keyword_file_path, 'r', encoding='utf-8') as file:
        keywords = set(line.strip() for line in file)

    # 创建一个字典来跟踪每个关键词的命中次数
    keyword_count = {keyword: 0 for keyword in keywords}
    seen = set()

    # 准备输出文件
    with open(output_file_path, 'w', encoding='utf-8') as output:
        # 逐行读取文本文件
        with open(text_file_path, 'r', encoding='utf-8') as text:
            for line_number, line in enumerate(text, start=1):
                # 检查当前行是否包含关键词
                for keyword in keywords:
                    if keyword in line:
                        # 增加关键词的命中次数
                        keyword_count[keyword] += 1
                        if line not in seen:
                            seen.add(line)
                            if get_match:
                            # 打印行号和行内容到输出文件
                                output.write((keyword+line).strip() + "###0\n")
                if line not in seen:
                    seen.add(line)
                    if not get_match:
                        output.write(line.strip() + "###1\n")

    # 按照命中次数从多到少排序并打印每个关键词的命中次数
    sorted_keyword_count = sorted(keyword_count.items(), key=lambda item: item[1], reverse=True)
    for keyword, count in sorted_keyword_count:
        print(f"Keyword '{keyword}' found {count} times.")


```

```python
import re


def filter_lines_by_regex(file_A_path, file_B_path, output_file_path, get_match=True):
    # 读取文件B中的正则表达式或普通符号
    with open(file_B_path, 'r', encoding='utf-8') as file_B:
        regex_patterns = [line.strip() for line in file_B.readlines() if line.strip()]

    # 编译正则表达式模式
    regex_compiled = [re.compile(pattern) for pattern in regex_patterns]

    with open(file_A_path, 'r', encoding='utf-8') as file_A, open(output_file_path, 'w', encoding='utf-8') as output_file:
        for line in file_A:
            # 对文件A的每一行文本进行正则表达式匹配
            if any(regex.search(line) for regex in regex_compiled):
                if get_match:
                    output_file.write(line.strip() + "###0\n")
            elif not get_match:
                output_file.write(line)

```

```python
import random

def random_lines_from_file(file_path, num_lines=1000):
    # 读取文件中的所有行
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    
    # 如果文件中的行数小于num_lines，则选择所有行
    if len(lines) < num_lines:
        selected_lines = lines
    else:
        # 否则，随机选择num_lines行
        selected_lines = random.sample(lines, num_lines)
    
    # 在每行末尾添加"###0"并返回结果列表
    return [line.strip() + '###0' for line in selected_lines]

def combine_random_lines(file_paths, output_file_path, num_lines=1000):
    # 存储从每个文件中随机选择的行
    combined_lines = []
    
    # 从每个文件中随机选择num_lines行
    for file_path in file_paths:
        lines = random_lines_from_file(file_path, num_lines)
        combined_lines.extend(lines)
    
    # 将合并后的行写入输出文件
    with open(output_file_path, 'w', encoding='utf-8') as output_file:
        output_file.write('\n'.join(combined_lines))


# 文件路径
file1 = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1headline_output.txt'
file2 = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1keyword_output.txt'
file3 = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1regax_output.txt'
output_file = '/home/zlj/code/pcl/template/data/middle_dataset/3neg_data/nega.txt'

# 调用函数
combine_random_lines([file1, file2, file3], output_file)
```

```python
def split_lines_by_length(file_A_path, file_B_path, file_C_path):
    # 打开文件A进行读取
    with open(file_A_path, 'r', encoding='utf-8') as file_A:
        lines = file_A.readlines()

    # 打开文件B和文件C准备写入
    with open(file_B_path, 'w', encoding='utf-8') as file_B, open(file_C_path, 'w', encoding='utf-8') as file_C:
        for line in lines:
            line = line.strip()  # 去除行尾的换行符
            if len(line) > 100:
                file_B.write(line + '\n')  # 写入文件B
            elif len(line) < 100:
                file_C.write(line + '\n')  # 写入文件C

# 调用函数
file_A_path = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/2positive_data.txt'  # 文件A的路径
file_B_path = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/3positive_data.txt'  # 文件B的路径
file_C_path = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/3short_positive_data.txt'  # 文件C的路径
split_lines_by_length(file_A_path, file_B_path, file_C_path)
```

```python
import random
import os

def random_lines_from_files(file_paths, output_file_path, lines_per_file):
    # 确保文件路径列表不为空且输出文件路径已指定
    if not file_paths or not output_file_path or len(file_paths) != 3:
        raise ValueError("需要提供三个文件路径和一个输出文件路径")
    
    # 计算总行数
    total_lines = lines_per_file * 3
    
    # 读取文件并随机选择行
    lines = []
    for file_path in file_paths:
        with open(file_path, 'r', encoding='utf-8') as file:
            # 读取所有行到内存中
            all_lines = file.readlines()
            # 随机选择指定数量的行
            selected_lines = random.sample(all_lines, min(lines_per_file, len(all_lines)))
            lines.extend(selected_lines)
    
    # 确保我们没有超过所需的总行数
    if len(lines) > total_lines:
        lines = random.sample(lines, total_lines)
    
    # 将选定的行写入新文件
    with open(output_file_path, 'w', encoding='utf-8') as output_file:
        output_file.writelines(lines)

# 使用示例
file_paths = ['/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1headline_output.txt',
               '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1keyword_output.txt',
                 '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/1regax_output.txt']
output_file_path = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/3negative.txt'
lines_per_file = 40000  # 每个文件选取的行数
random_lines_from_files(file_paths, output_file_path, lines_per_file)
```
