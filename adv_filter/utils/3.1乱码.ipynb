```python
import re

def find_non_chinese_lines(file_path):
    # 定义一个正则表达式模式，匹配20个以上连续的非中文字符
    pattern = re.compile(r'[^\u4e00-\u9fff]+{20,}')

    # 打开文件，读取每一行，并检查是否匹配模式
    with open(file_path, 'r', encoding='utf-8') as file:
        for line_number, line in enumerate(file, 1):
            if pattern.search(line):
                print(f'Line {line_number}: {line.strip()}')  # 打印匹配的行

# 使用函数
# 请将'yourfile.txt'替换为你的文件名
find_non_chinese_lines('/home/zlj/code/pcl/template/data/middle_dataset/2remove/filtered1000.txt')
```
