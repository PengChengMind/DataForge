```python
def search_keywords(text_file_path, keyword_file_path, output_file_path, get_match=True):
    # 读取关键词库
    with open(keyword_file_path, 'r', encoding='utf-8') as file:
        keywords = set(line.strip() for line in file)

    # 创建一个字典来跟踪每个关键词的命中次数
    keyword_count = {keyword: 0 for keyword in keywords}
    seen = set()

    # 准备输出文件
    with open(output_file_path, 'w', encoding='utf-8') as output:
        # 逐行读取文本文件
        with open(text_file_path, 'r', encoding='utf-8') as text:
            for line_number, line in enumerate(text, start=1):
                # 检查当前行是否包含关键词
                for keyword in keywords:
                    if keyword in line:
                        # 增加关键词的命中次数
                        keyword_count[keyword] += 1
                        if line not in seen:
                            seen.add(line)
                            if get_match:
                            # 打印行号和行内容到输出文件
                                output.write(line.strip() + "###0\n")
                if line not in seen:
                    seen.add(line)
                    if not get_match:
                        output.write(line.strip() + "###1\n")

    # 按照命中次数从多到少排序并打印每个关键词的命中次数
    sorted_keyword_count = sorted(keyword_count.items(), key=lambda item: item[1], reverse=True)
    for keyword, count in sorted_keyword_count:
        print(f"Keyword '{keyword}' found {count} times.")
```

```python
# 指定文件路径
text_file_path = '/home/zlj/code/pcl/template/data/middle_dataset/2remove/filtered1000.txt'  # 3万行文本文件路径
keyword_file_path = '/home/zlj/code/pcl/template/data/keywords/keywordForNeg.txt'  # 1500行关键词库文件路径，更加严格
output_file_path = '/home/zlj/code/pcl/template/data/middle_dataset/3keyword/filtered1000_neg.txt'  # 输出文件路径

# 调用函数
search_keywords(text_file_path, keyword_file_path, output_file_path, True)
```
