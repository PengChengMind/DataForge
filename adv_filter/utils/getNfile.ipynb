```python
import os
import random

def read_and_write_files(source_dir, output_file_path, num):
    # 确保输出文件是空的
    output_dir = output_file_path+'unfiltered'+str(num)+".txt"

    open(output_dir, 'w').close()

    # 获取目录中的所有文件
    files = [f for f in os.listdir(source_dir) if os.path.isfile(os.path.join(source_dir, f))]

    random_numbers = random.sample(range(0, len(files)), num)
    # 限制只处理前num个文件
    files = [files[num] for num in random_numbers]
    # 遍历文件列表
    for file_name in files:
        file_path = os.path.join(source_dir, file_name)
        with open(file_path, 'r', encoding='utf-8') as file:
            # 读取文件的每一行
            for line in file:
                # 写入到输出文件
                with open(output_dir, 'a', encoding='utf-8') as out_file:
                    out_file.write(line)

```

```python
# 指定源目录和输出文件路径
source_directory = '/data/zlj/pcl_data/CC-MAIN-10G'
output_file_path = '/home/zlj/code/pcl/template/data/ori_dataset/CC-MAIN/'

# 调用函数
read_and_write_files(source_directory, output_file_path, 1000)
```
