```python
import pandas as pd

# 指定Excel文件路径
excel_file_path = '/home/zlj/code/pcl/template/data/dealdata/testdata.xlsx'
# 指定输出文本文件路径
output_txt_file_path = '/home/zlj/code/pcl/template/data/test_data/test.txt'

# 读取Excel文件，从第二行开始读（跳过第一行）
df = pd.read_excel(excel_file_path)

# 打开文件准备写入
with open(output_txt_file_path, 'w', encoding='utf-8') as output_file:
    # 遍历DataFrame中的每一行
    for index, row in df.iterrows():
        data = row['Text'].replace('\\n','').replace('\n','')  # First column (index 0) is the test data
        label = 1-row['10包含广告']  # Fourth column (index 3) is the label
        output_file.write(f'{data}###{label}\n')

print("文件已成功创建")
```

```python
import pandas as pd

def save_second_column_to_file(xlsx_file_path, output_file_path, add_suffix):
    """
    读取xlsx文件的第二列，跳过首行，将每行内容存到指定文件中，并根据参数添加后缀。

    :param xlsx_file_path: str, xlsx文件的路径
    :param output_file_path: str, 输出文件的路径
    :param add_suffix: int, 根据此参数值在行末添加'###0'或'###1'
    """
    # 使用pandas读取xlsx文件
    df = pd.read_excel(xlsx_file_path)
    
    # 获取第二列的数据，跳过首行
    
    # 打开输出文件准备写入
    with open(output_file_path, 'w', encoding='utf-8') as file:
        # 遍历第二列的数据
        for index, row in df.iterrows():
            item = row['样本数据'].replace('\\n','').replace('\n','')
            # 根据参数添加后缀
            if add_suffix == 0:
                file.write(f"{item}###0\n")
            elif add_suffix == 1:
                file.write(f"{item}###1\n")
            else:
                raise ValueError("add_suffix参数只能是0或1")

# 使用示例
# save_second_column_to_file('path_to_your_xlsx_file.xlsx', 'output.txt', 0)
```
