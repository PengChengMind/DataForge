```python
# 打开文件并读取所有行
with open('/home/zlj/code/pcl/template/data/keywords/keywordForNeg.txt', 'r') as file:
    lines = file.readlines()

# 移除每行的首尾空白字符（包括换行符），然后对行进行排序
sorted_lines = sorted([line.strip() for line in lines])

# 将排序后的行写回文件
with open('/home/zlj/code/pcl/template/data/keywords/keywordForNeg_sort.txt', 'w') as file:
    for line in sorted_lines:
        file.write(line + '\n')
```
