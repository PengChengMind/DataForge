```python
# 打开文件
with open('/home/zlj/code/pcl/template/data/undealdata/negative.txt', 'r') as file:
    lines = file.readlines()

# 处理每一行
processed_lines = []
for line in lines:
    # 去除行尾的换行符
    line = line.rstrip('\n')
    # 保留最后5个字符，其余部分删除'#'
    processed_line = line[:-5].replace('#', '') + line[-5:]
    processed_lines.append(processed_line)

# 将处理后的内容写回文件或进行其他操作
with open('/home/zlj/code/pcl/template/data/dealdata/negative.txt', 'w') as file:
    for processed_line in processed_lines:
        file.write(processed_line + '\n')
```
