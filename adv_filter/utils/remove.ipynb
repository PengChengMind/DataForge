```python

def remove_re(input_file_path, output_file_path):
# 使用集合来存储已经出现过的行
    seen = set()
    with open(input_file_path, 'r') as file, open(output_file_path, 'w') as output:
        for line in file:
            if line not in seen:
                seen.add(line)
                output.write(line)
```

```python
def remove_and_append(file_input_path, file_output_path):
    # 打开输入文件，读取内容并去除行尾的 ###0 或 ###1
    with open(file_input_path, 'r', encoding='utf-8') as file_input:
        lines = [line.rstrip() for line in file_input]  # 去除行末空白符

    # 去除行尾的 ###0 或 ###1
    processed_lines = [line[:-4] if line.endswith('###0') or line.endswith('###1') else line for line in lines]

    # 将处理后的内容附加到现有文件的底部
    with open(file_output_path, 'a', encoding='utf-8') as file_output:  # 'a' 模式用于附加内容
        for line in processed_lines:
            file_output.write(line + '\n')

# 使用示例
file_input = '/home/zlj/code/pcl/template/data/middle_dataset/2remove/oridata2.txt'
file_output = '/home/zlj/code/pcl/template/data/middle_dataset/2.5makeup_data/filtered1000.txt'

remove_and_append(file_input, file_output)
```
