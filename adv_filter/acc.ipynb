```python
import pandas as pd

# 指定Excel文件的路径
excel_file_path = '/home/zlj/code/pcl/template/data/dealdata/testdata.xlsx'

# 使用pandas的read_excel函数读取数据，从第二行开始读取（skiprows=1）
df = pd.read_excel(excel_file_path, skiprows=1)

# 假设第二列是预测标签，第四列是真实标签
# 列的索引从0开始，所以第二列是索引1，第四列是索引3
predicted_labels = 1 - df.iloc[:, 1]  # 第二列
true_labels = df.iloc[:, 3]  # 第四列

# 计算预测准确率
accuracy = (predicted_labels == true_labels).mean()  # 计算两列相等的比例

# 打印准确率
print(f'预测准确率为: {accuracy:.2f}')
```
